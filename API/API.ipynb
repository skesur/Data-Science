```python
import requests
import json
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "notebook"
```

```python
lat=23.033863
lon=72.585022
key="57ed3a63cbe79efa475d0ab0b2f29b27"
url=f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={key}"
data=requests.get(url)
data.status_code
```

```python
new=data.json()
final=json.dumps(new,indent=5)
print(final)
```

```python
df=pd.DataFrame({"name":[new["name"]],"timezone":[new["timezone"]],"country":[new["sys"]["country"]],"wind speed":[new["wind"]["speed"]],
                 "visibility":[new["visibility"]],"description":[new["weather"][0]["description"]]})
df
```

```python
lat=23.033863
lon=72.585022
key="57ed3a63cbe79efa475d0ab0b2f29b27"
url=f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&appid={key}"
data=requests.get(url)
data.status_code
```

```python
new=data.json()
final=json.dumps(new,indent=5)
print(final)
```

```python
temp=[]
date=[]
time=[]
dis=[]
for i in new["list"]:
    temp.append(i["main"]["temp"])
    date.append(i["dt_txt"].split(" ")[0])
    time.append(i["dt_txt"].split(" ")[1])
    dis.append(i["weather"][0]["description"])    
df=pd.DataFrame({"temp":temp,"date":date,"time":time,"dis":dis})
df
```

```python
def cel(tem):
    return tem-273.15
df["temp_c"]=df["temp"].apply(cel)
df
```

```python
fig = px.bar(df[df["time"]=="12:00:00"],x="date",y="temp_c")
fig.show()
```
