```python
import requests
import json
import pandas as pd
```

```python
d={
    "Ahmedavad":("23.033863","72.585022"),
    "Mumbai":("19.076090","72.877426"),
    "Chennai":("13.0843","80.2705"),
    "Delhi":("28.7041","77.1025"),
    "Bengaluru":("12.9716","77.5946")
}
key="57ed3a63cbe79efa475d0ab0b2f29b27"
co=[]
no=[]
no2=[]
o3=[]
so2=[]
pm2_5=[]
pm10=[]
nh3=[]
for i in d:
    lat=d[i][0]
    lon=d[i][1]
    url=f"http://api.openweathermap.org/data/2.5/air_pollution?lat={lat}&lon={lon}&appid={key}"
    data=requests.get(url)
    new=data.json()
    for j in new["list"][0]["components"]:
        if j=="co":
            co.append(new["list"][0]["components"][j])
        elif j=="no":
            no.append(new["list"][0]["components"][j])
        elif j=="no2":
            no2.append(new["list"][0]["components"][j])
        elif j=="o3":
            o3.append(new["list"][0]["components"][j])
        elif j=="so2":
            so2.append(new["list"][0]["components"][j])
        elif j=="pm2_5":
            pm2_5.append(new["list"][0]["components"][j])
        elif j=="pm10":
            pm10.append(new["list"][0]["components"][j])
        elif j=="nh3":
            nh3.append(new["list"][0]["components"][j])
    print(data.status_code)    
df=pd.DataFrame({"co":co,"no":no,"no2":no2,"o3":o3,"so2":so2,"pm2_5":pm2_5,"pm10":pm10,"nh3":nh3},index=list(d.keys()))
df
```
