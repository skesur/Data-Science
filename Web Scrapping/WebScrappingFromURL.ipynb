```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

```python
url="https://www.politifact.com/factchecks/list/"
data=requests.get(url)
html=data.text
data.status_code
```

```python
soup=BeautifulSoup(html,"html.parser")
```

```python
x=soup.find_all("li",class_="o-listicle__item")
name = []
state = []
date = []
author = []
for i in x:
    n=i.find("a",class_="m-statement__name").text.strip()
    name.append(n)
    s=i.find("div",class_="m-statement__quote").text.strip()
    state.append(s)
    d=i.find("footer",class_="m-statement__footer").text.strip().split("•")[1]
    date.append(d)
    a=i.find("footer",class_="m-statement__footer").text.strip().split("•")[0]
    author.append(a)
df=pd.DataFrame({"name":name,"state":state,"date":date,"author":author})
df
```

```python
url="https://www.cricbuzz.com/cricket-series/9241/indian-premier-league-2026/points-table"
data=requests.get(url)
html=data.text
data.status_code
```

```python
soup=BeautifulSoup(html,"html.parser")
```

```python
x = soup.find_all("div",class_="grid point-table-grid p-2 cursor-pointer")
name=[]
played=[]
won=[]
lost=[]
nr=[]
pts=[]
nrr=[]
n=soup.find_all("div",class_="flex flex-row items-center gap-1")
for i in n:
    n1=i.find("span",class_="text-gray-600").text
    name.append(n1)
for i in x:
    a=i.find_all("div",class_="flex justify-start items-center text-sm text-cbPointsTableTextGrey")
    for j in a[2]:
        played.append(j.text)
    for j in a[3]:
        won.append(j)
    for j in a[4]:
        lost.append(j)
    for j in a[5]:
        nr.append(j)
    for j in a[6]:
        nrr.append(j)
s=soup.find_all("div",class_="flex justify-start items-center text-sm font-bold text-gray-800")
for i in s:
    pts.append(i.text)

played=played[1:]
won=won[1:]
lost=lost[1:]
nr=nr[1:]
nrr=nrr[1:]
df=pd.DataFrame({"TEAMS":name,"P":played,"W":won,"L":lost,"NR":nr,"PTS":pts,"NRR":nrr})
df
```
