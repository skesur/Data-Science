```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("400.html","r",encoding="utf-8") as f:
     data = f.read()
soup = BeautifulSoup(data,"html.parser")
```

```python
x=soup.find_all("div",class_="job_seen_beacon")
n=[]
c=[]
a1=[]
a2=[]
s=[]
for i in x:
    name=i.find("div",class_="css-pt3vth e37uo190").text
    n.append(name)
    com=i.find("span",class_="css-1h7lukg eu4oa1w0").text
    c.append(com)
    add=i.find("div",class_="css-1restlb eu4oa1w0").text
    a1.append(add.split(", ")[0])
    a2.append(add.split(", ")[1])
    sal=i.find("div",class_="css-18z4q2i eu4oa1w0")
    if sal==None:
        s.append("Not Defined")
    else:
        if "₹" in sal.text:
            s.append(sal.text)
        else:
            s.append("Not Defined")
df=pd.DataFrame({"Role":n,"Company":c,"City":a1,"State":a2,"Salary":s})
df
```
