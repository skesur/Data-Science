```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("beautifulsoup_practice.html","r") as f:
     data = f.read()
```

```python
soup = BeautifulSoup(data,"html.parser")
```

```python
t=soup.find("title").text
n=soup.find("h2").text
d=soup.find("p",{"class":"department"}).text
e=soup.find("span").text
l=soup.find("a")["href"]
i=soup.find("img")["src"]
ext=e.split("@")[1]
```

```python
data = {"title":[t],"name":[n],"dep":[d],"email":[e],"link":[l],"img":[i],"extension":[ext]}
df=pd.DataFrame(data)
df
```

```python
x=soup.find_all("div",{"class":"student"})
n=[]
d=[]
e=[]
l=[]
i1=[]
ext=[]
for i in x:
    name=i.find("h2").text
    n.append(name)
    dep=i.find("p",{"class":"department"}).text
    d.append(dep)
    email=i.find("span").text
    e.append(email)
    link=i.find("a")["href"]
    l.append(link)
    img=i.find("img")["src"]
    i1.append(img)
    exte=email.split("@")[1]
    ext.append(exte)
 
data = {"name":n,"dep":d,"email":e,"link":l,"img":i1,"extension":ext}
df=pd.DataFrame(data)
df.to_csv("beautifulsoup_practice_1.csv")
```
