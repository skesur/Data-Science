```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("flipkart.html","r",encoding="utf-8") as f:
     data = f.read()
soup = BeautifulSoup(data,"html.parser")    
```

```python
x=soup.find_all("div",{"class":"_3pLy-c row"})
n=[]
r=[]
b=[]
for i in x:
    name=i.find("div",{"class":"_4rR01T"}).text
    n.append(name)
    rate=i.find("div",{"class":"_3LWZlK"}).text
    r.append(rate)
    batt=i.find_all("li",{"class":"rgWa7D"})
    ba=""
    for j in batt:
        if "Battery" in j.text:
            ba=j.text;
            break;
    b.append(ba)
df=pd.DataFrame({"name":n,"rating":r,"battery":b})
df
```
