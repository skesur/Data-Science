```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("flipkart.html","r",encoding="utf-8") as f:
     data = f.read()
soup = BeautifulSoup(data,"html.parser")    
```

```python
x=soup.find_all("div",{"class":"_3pLy-c row"})
n=[]
r=[]
s=[]
b=[]
d=[]
o=[]
for i in x:
    name=i.find("div",{"class":"_4rR01T"}).text
    n.append(name)
    rate=i.find("div",{"class":"_3LWZlK"}).text
    r.append(rate)
    sto=i.find("li",{"class":"rgWa7D"}).text
    s.append(sto)
    batt=i.find_all("li",{"class":"rgWa7D"})
    ba=""
    for j in batt:
        if "Battery" in j.text:
            ba=j.text;
            break;
    b.append(ba)
    dis=i.find("div",{"class":"_30jeq3 _1_WHN1"}).text
    d.append(dis)
    ori=i.find("div",{"class":"_3I9_wc _27UcVY"})
    if ori:
        ori=ori.text
    else:
        ori=0
    o.append(ori)    
    
df=pd.DataFrame({"name":n,"rating":r,"storage":s,"battery":b,"discount":d,"origin":o})
df
```

```python
num=[]
for i in d:
    i=i[1:]
    i=i.replace(",","")
    i=int(i)
    num.append(i)
df["discount"]=num    
```

```python
num1=[]
for i in o:
    if i!=0:
        i=i[1:]
        i=i.replace(",","")
        i=int(i)
        num1.append(i)
    else:
        num1.append(0)
df["origin"]=num1  
```

```python
df
```

```python
df["price"]=df["origin"]-df["discount"]
df["price"]=df["price"]/df["origin"]
df
```

```python
import numpy as np
def per(price):
    if price==-np.inf:
        return "No Discount"
    else:
        return int(price*100)
df["price"]=df["price"].apply(per)
df
```

```python
rom=[]
ram=[]
for i in s:
    i1=i.split("|")
    ra=""
    ro=""
    if "RAM" in i1[0]:
        ra=i1[0]
    else :
        ra="NO RAM"
    if len(i1)!=1:
        if "ROM" in i1[1]:
            ro=i1[1]
        else :
            ro="NO ROM"
    else :
        if "ROM" in i1[0]:
            ro=i1[0]
        else :
            ro="NO ROM"
    ram.append(ra)
    rom.append(ro)
df["rom"]=rom
df["ram"]=ram
df
```
