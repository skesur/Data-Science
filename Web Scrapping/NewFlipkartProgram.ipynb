```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open("newflipkart.html","r",encoding="utf-8") as f:
    data = f.read()
soup=BeautifulSoup(data)
```

```python
x = soup.find_all("div",class_="yKfJKb row")
na=[]
m=[]
c=[]
s=[]
r=[]
op=[]
np=[]
d=[]
e=[]
for i in x:
    name=i.find("div",class_="KzDlHZ")
    na.append(name.text.split(" (")[0])
    c1 = name.text.split(" (")[1]
    c.append(c1.split(",")[0])
    star=i.find("div",class_="XQDdHH")
    s.append(star.text)
    memory=i.find_all("li",class_="J+igdf")
    rating=i.find("span",class_="Wphh3N")
    r.append(rating.text.split("&")[1].split(" ")[0])
    old=i.find("div",class_="yRaY8j ZYYwLA")
    op.append(old.text)
    new=i.find("div",class_="Nx9bqj _4b5DiR")
    np.append(new.text)
    dis=i.find("div",class_="UkUFwK")
    d.append(dis.text.split(" ")[0])
    ex=i.find_all("div",class_="yiggsN O5Fpg8")
    if len(ex)==4:
        e1=ex[2].text.replace(",","")
        e1=e1.replace("₹","")
        e.append(int(e1))
    else :
        e.append("Not defined")
    for j in memory:
        if "ROM" in j.text:
            m.append(j.text)

op11=[]
for i in op:
    o=i.replace("₹","")
    o=o.replace(",","")
    op11.append(int(o))
np11=[]
for i in np:
    n=i.replace("₹","")
    n=n.replace(",","")
    np11.append(int(n))
df=pd.DataFrame({"Name":na,"Memory":m,"Color":c,"Star":s,"Review":r,"Old Price":op11,"New Price":np11,"Discount":d,"Exchange Value":e})
df["Difference"]=df["Old Price"]-df["New Price"]
df
```
