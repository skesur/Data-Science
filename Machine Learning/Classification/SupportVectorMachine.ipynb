```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
```

```python
df = pd.DataFrame(pd.read_csv("cancer.csv"));
df.drop(["id","Unnamed: 32"],axis=1,inplace=True)
df
```

```python
x = df.drop(["diagnosis"],axis=1)
y = df["diagnosis"]
x = pd.get_dummies(x)
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
model = SVC(kernel="rbf",C=5,random_state=42)
model.fit(x_train,y_train)
pre = model.predict(x_test)
print(accuracy_score(y_test,pre))
print(classification_report(y_test,pre))
print(confusion_matrix(y_test,pre))
```

```python
sns.heatmap(confusion_matrix(y_test,pre),annot=True)
```

```python
sns.scatterplot(x=y_test, y=pre)
```

```python
dic = {}
for i in range(1,100):
    model = SVC(kernel="rbf",C=i,random_state=42)
    model.fit(x_train,y_train)
    pre = model.predict(x_test)
    dic[i] = accuracy_score(y_test,pre)
max_key = max(dic,key=dic.get)
print(max_key,dic[max_key])
```
