```python
import pandas as pd
```

```python
df = pd.DataFrame(pd.read_csv("heights_weights.csv"))
df.drop(columns="Male",axis=1,inplace=True)
df
```

```python
x=df[["Height"]]
y=df["Weight"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
for i in range(6):
    x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3,random_state=(i+1))
    model = LinearRegression()
    z = model.fit(x_train,y_train)
    p = model.predict(x_test)
    r = r2_score(y_test,p)
    m = mean_squared_error(y_test,p)
    print(i+1)
    print(r,m)
    print(model.coef_,model.intercept_)
    sns.scatterplot(x=y_test,y=p)
```
